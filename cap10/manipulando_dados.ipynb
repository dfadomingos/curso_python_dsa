### Preenchendo valores ausentes

```python
#a função fillna() é usada para preencher valores ausentes
dsa_df = pd.read_csv("dataset.csv")
```

```python
dsa_df.head(5)
```

```python
dsa_df.isna().sum()
```

```python
#extraimos a moda da coluna Quantify
moda = dsa_df['Quantidade'].value_counts().index[0]

print(moda)
```

```python
#por fim preenchendo os valores NA com a moda
dsa_df['Quantidade'].fillna(value = moda, inplace = True)
```

```python
dsa_df.isna().sum()
```

### Query(consulta) de Dados no DataFrame do Pandas

```python
#gerando um novo dataframe apenas com o intervalo de vendas entre 229 e 10000
df2 = dsa_df.query('229 < Valor_Venda < 10000')
```

```python
#confirmando os valores maximos e mininos
df2.Valor_Venda.describe()
```

```python
#gerando um novo dataframe apenas com valores acima da media
df3 = df2.query('Valor_Venda > 766')
```
