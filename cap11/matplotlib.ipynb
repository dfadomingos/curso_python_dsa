```python
import matplotlib.pyplot as plt
```

### Contruindo plots

```python
#o metodo plot define os eixos dos graficos
plt.plot([1, 3, 5], [2, 4, 7])
plt.show()
```

```python
x = [2, 3, 5]
y = [3, 5, 7]
```

```python
plt.plot(x,y)
plt.xlabel('Variavel 1')
plt.ylabel('Variavel 2')
plt.title('Teste Plot')
plt.show()
```

```python
x2 = [1, 2, 3]
y2 = [11, 12, 15]
```

```python
plt.plot(x2, y2, label = 'Gráfico com matplotlib')
plt.legend()
plt.show()
```
