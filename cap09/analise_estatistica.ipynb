```python
import numpy as np
```

```python
arr1 = np.array([15, 23, 63, 94, 75])
```

```python
#media
np.mean(arr1)
```

```python
#desvio padrao
np.std(arr1)
```

```python
#variancia
#medida para avaliar a variabilidade dos dados em torno da media.
#se for baixa, indica que os valores estão próximos da media
#se for alta, indica que os valores estão distantes da media e tem alta variabilidade
np.var(arr1)
```
