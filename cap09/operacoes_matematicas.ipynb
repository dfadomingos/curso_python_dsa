```python
import numpy as np
```

```python
arr1 = np.arange(1, 10)
print(arr1)
```

```python
#soma dos elementos
np.sum(arr1)
```

```python
#produto
np.prod(arr1)
```

```python
#soma acumulada dos elementos
np.cumsum(arr1)
```

```python
arr2 = np.array([3, 2, 1])
arr3 = np.array([1, 2, 3])
```

```python
#soma dos array
arr4 = np.add(arr2, arr3)
```

```python
print(arr4)
```

```python
#para multiplicar matrizes podemos usar a função dot() ou o operador @
arr5 = np.array([[1, 2], [3, 4]])
arr6 = np.array([[5, 6], [0, 7]])
```

```python
print(arr6)
```

```python
#multiplicação
arr7 = np.dot(arr5, arr6)
print(arr7)
```

```python
arr8 = arr5 @ arr6
print(arr8)
```
