```python
#comparação global
arr2 = np.arange(10)

np.array_equal(arr1, arr2)
```

```python
#somando um valor a cada elemento do array
arr3 = np.array([1, 2, 3]) +1.5
print(arr3)
```

```python
#arredondando
np.around(arr3)
```

```python
arr4 = np.array([[1, 2, 3, 4], [5, 6, 7, 8]])
print(arr4)
```

```python
#metodo flatten é usado para criar uma copia unidimensional de um array multidimensional
arr5 = arr4.flatten() 
print(arr5)
```
