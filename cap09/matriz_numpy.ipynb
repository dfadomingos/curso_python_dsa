```python
import numpy as np
```

```python
arr1 = np.array([[1,2,3], [4,5,6]])
print(arr1)
```

```python
print(arr1.shape)
```

```python
#criando uma lista 2x3 apenas com numeros 1
arr2 = np.ones((2,3))
print(arr2)
```

```python
#lista de listas
lista = [[13,81,22], [0,34,59], [21,48,94]]
```

```python
#a funcao matrix cria uma matriz a partir de uma lista de listas
arr3 = np.matrix(lista)
print(arr3)
```

```python
print(arr3.size)
```

```python
#indexação da matriz (linha, coluna)
arr3[2,1]
```

```python
arr3[0:2,2]
```

```python
arr3[1,]  #linha 1
```

```python
#alterando elemento da matriz
arr3[1,0] = 100
print(arr3)
```

```python
#itemsize de um array numpy é um atributo que retorna o tamanho em bytes de cada elemento do array. O itemsize representa o numero de bytes necessarios para armazenar cada valor do array numpy

arr4 = np.array([[24,76,92,14], [47,35,89,2]], dtype=float)
print(arr4)
```

```python
arr4.itemsize
```

```python
arr4.nbytes
```

```python
arr4.ndim   #dimensoes
```
