```python
import numpy as np
```

```python
#a função array cria um array numpy
arr2 = np.array([1, 3, 6, 10, 15])
```

```python
print(arr2)
```

```python
#usando metodos do array
arr2.cumsum()
```

```python
arr2.cumprod()
```

```python
arr3 = np.arange(0, 50, 5)
print(arr3)
```

```python
np.shape(arr3)
```

```python
#cria um array preenchido com zeros
arr4 = np.zeros(10)
arr4
```

```python
#retorna 1 nas posiçoes em diagonal e 0 no restante
#matriz diagonal
arr5 = np.eye(4)
arr5
```

```python
#os valores passados como parametro, forma uma diagonal
arr6 = np.diag(np.array([1, 2, 3, 4]))
arr6
```

```python
#array com valores booleanos
arr7 = np.array([True, False, True, False])
arr7
```

```python
#array de strings
arr8 = np.array(['Lingagem Python', 'Linguagem R', 'Linguagem C'])
print(arr8)
```

```python
#A funçao linspace do numpy é usada para criar uma sequencia de numeros igualmente espaçados dentro de um intervalo especifico
print(np.linspace(0 , 10))
print('\n')
print(np.linspace(0, 10, 15))
```

```python
#a funcao logspace é usada para criar uma sequencia de numeros igualmente espaçados em escala logaritma dentro de um intervalo especifico
print(np.logspace(0, 5, 10))
```
