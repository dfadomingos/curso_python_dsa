```python
import numpy as np
```

### Criando array

```python
#array numpy muito mais eficiente, veloz e versatil que uma lista normal

arr1 = np.array([10, 21, 32, 43, 48, 15, 76, 57, 89])
```

```python
print(arr1)
```

```python
arr1.shape

#shape 9, > 1D , shape 2,3 >> 2D, shape 4,3,2 >> 3D
```

### Indexação em arrays numpy

```python
print(arr1)
```

```python
arr1[4]
```

```python
arr1[1:4]
```

```python
arr1[1:4+1]
```

```python
#cria uma lista de indices
indices = [1, 2, 5, 6]
```

```python
arr1[indices]
```

```python
#mascara booleana para os elementos pares
mask = (arr1 % 2 == 0)
```

```python
mask
```

```python
arr1[mask]
```

```python
#alterando um elemento do array
arr1[0] = 100
print(arr1)
```

```python
#nao é possivel incluir elemento de outro tipo
try:
    arr1[0] = 'Novo elemento'

except:
    print('Operação não permitida')
```
